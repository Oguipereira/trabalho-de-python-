```python
# Tarefa 1: diz se o primeiro número é múltiplo do segundo

a = int(input("Digite o primeiro número: "))
b = int(input("Digite o segundo número: "))

if b == 0:
    multiplo = (a == 0)      # só o zero é múltiplo de zero (evita divisão por zero)
else:
    multiplo = (a % b == 0)  # resto zero => é múltiplo

if multiplo:
    print(f"{a} é múltiplo de {b}")
else:
    print(f"{a} não é múltiplo de {b}")
```

```python
# Tarefa 2: escreve os múltiplos de 5 entre 23 e 1502

n = 23
while n <= 1502:
    if n % 5 == 0:
        print(n)
    n = n + 1
```
